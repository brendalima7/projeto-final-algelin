# Video Background Removal Using Singular Value Decomposition (SVD)

```python
import os
import sys
import cv2
import numpy as np

external_path = os.path.abspath(
    os.path.join(os.getcwd(), '..')
)

sys.path.append(external_path)

from utils import (
    build_matrix,
    column_to_image,
    load_video,
    separate_background,
    to_uint8,
    save_video
)
```

```python
n_video = 2
```

```python
video_path = f"../videos/original/video_0{n_video}.mp4"
true_background = None

frames = load_video(video_path)
M, shape = build_matrix(frames)
background, residual, s = separate_background(M, k=1)
```

```python
save_video([column_to_image(background[:, t], shape) for t in range(M.shape[1])],
           f"../videos/background/background_0{n_video}.mp4")

cv2.imwrite(f"../images/svd/background/background_0{n_video}.png",
            to_uint8(column_to_image(np.median(background, axis=1), shape)))
```
